```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("File_Formats") \
    .master("local[*]") \
    .getOrCreate()

spark
```

```python
csv_file_with_schema = spark.read.format("csv") \
.option("header", "true") \
.option("inferSchema", "true") \
.load("data/titanic.csv")

csv_file_with_schema.show()
csv_file_with_schema.printSchema()
```

### JSON

```python
json_file = spark.read.format("json").load("data/titanic.json")
json_file.show()
json_file.printSchema()
```

### Parquet

```python
csv_file_with_schema.write.mode("overwrite").format("parquet").save("titanic_parquet/")
```

```python
df_parquet = spark.read.format("parquet").load("titanic_parquet")

df_parquet.show()
df_parquet.printSchema()
```
